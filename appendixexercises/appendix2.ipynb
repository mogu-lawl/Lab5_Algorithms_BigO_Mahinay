```python
import random
import time

print("=" * 60)
print(" STUDENT-SPECIFIC INPUTS & DERIVED VALUES ")
print("=" * 60)

# Student-Specific Base Inputs (Modify these with your actual details if needed)
LAST_NAME = "Mahinay"
STUDENT_ID = "TUPM-26-4014"
FAVORITE_ARTIST = "THE 1975"

# Derived Values based on Input Rules
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
# Sum of A=1 to Z=26 values for FAVORITE_ARTIST (ignoring spaces)
ARTIST_SUM = sum(ord(char.upper()) - ord('A') + 1 for char in FAVORITE_ARTIST if char.isalpha())

print(f"LAST_NAME: {LAST_NAME}")
print(f"STUDENT_ID: {STUDENT_ID}")
print(f"FAVORITE_ARTIST: {FAVORITE_ARTIST}")
print(f"SEED_DIGIT: {SEED_DIGIT}")
print(f"ID_SUM: {ID_SUM}")
print(f"NAME_LENGTH: {NAME_LENGTH}")
print(f"ARTIST_LENGTH: {ARTIST_LENGTH}")
print(f"ARTIST_SUM: {ARTIST_SUM}\n")

# =====================================================================
# 2. UNIVERSITY ENROLLMENT RECORDS
# =====================================================================
print("=" * 60)
print(" EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS ")
print("=" * 60)

# 1. Generate Department A and Department B datasets (sorted ascending)
dept_a = sorted([ID_SUM + i * (SEED_DIGIT + 2) for i in range(6)])
dept_b = sorted([ARTIST_LENGTH * 10 + i * (NAME_LENGTH + 1) for i in range(6)])

# 2. Consolidate (Merge two sorted lists)
def merge_sorted_lists(a, b):
    merged = []
    i = j = 0
    comparisons = 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    while i < len(a):
        merged.append(a[i])
        i += 1
    while j < len(b):
        merged.append(b[j])
        j += 1
    return merged, comparisons

consolidated_dataset, merge_comparisons = merge_sorted_lists(dept_a, dept_b)

# 3. Verification
records_before = len(dept_a) + len(dept_b)
records_after = len(consolidated_dataset)
is_ordered = all(consolidated_dataset[i] <= consolidated_dataset[i+1] for i in range(len(consolidated_dataset) - 1))

# Assessment Data Output - Exercise 2
print("Assessment Data:")
print(f"1. Department A dataset: {dept_a}")
print(f"2. Department B dataset: {dept_b}")
print(f"3. Consolidated dataset: {consolidated_dataset}")
print(f"4. Number of records before and after consolidation: Before = {records_before}, After = {records_after}")
print(f"5. Comparison count: {merge_comparisons}")
print(f"6. Ordering verification: {'Verified - Perfectly Sorted' if is_ordered else 'Failed'}")
print(f"7. Time and space complexity: Time: O(n + m), Space: O(n + m)")
print(f"8. Final Output: Successfully merged and verified {records_after} sorted enrollment records.\n")
```
