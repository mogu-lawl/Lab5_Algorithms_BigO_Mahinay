```python
import random
import time

print("=" * 60)
print(" STUDENT-SPECIFIC INPUTS & DERIVED VALUES ")
print("=" * 60)

# Student-Specific Base Inputs (Modify these with your actual details if needed)
LAST_NAME = "Mahinay"
STUDENT_ID = "TUPM-26-4014"
FAVORITE_ARTIST = "THE 1975"

# Derived Values based on Input Rules
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
# Sum of A=1 to Z=26 values for FAVORITE_ARTIST (ignoring spaces)
ARTIST_SUM = sum(ord(char.upper()) - ord('A') + 1 for char in FAVORITE_ARTIST if char.isalpha())

print(f"LAST_NAME: {LAST_NAME}")
print(f"STUDENT_ID: {STUDENT_ID}")
print(f"FAVORITE_ARTIST: {FAVORITE_ARTIST}")
print(f"SEED_DIGIT: {SEED_DIGIT}")
print(f"ID_SUM: {ID_SUM}")
print(f"NAME_LENGTH: {NAME_LENGTH}")
print(f"ARTIST_LENGTH: {ARTIST_LENGTH}")
print(f"ARTIST_SUM: {ARTIST_SUM}\n")


# =====================================================================
# 1. EMERGENCY DISPATCH RECORD SEARCH
# =====================================================================
print("=" * 60)
print(" EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH ")
print("=" * 60)

# 1. Generate Original Dataset (unique sorted values using student inputs)
base_value = ID_SUM * 10
original_dataset = [base_value + i * (SEED_DIGIT + 1) for i in range(12)]

# 2. Rotated Dataset (rotate by SEED_DIGIT)
rotation_index = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_index:] + original_dataset[:rotation_index]

# 3. Target Incident Generation
target_incident = original_dataset[len(original_dataset) // 2] # Ensure target exists in dataset

# 4. Rotated Binary Search Implementation & Comparison Count
def rotated_binary_search(arr, target):
    left, right = 0, len(arr) - 1
    comparisons = 0
    while left <= right:
        comparisons += 1
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid, comparisons
        
        # Check if left half is sorted
        if arr[left] <= arr[mid]:
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else: # Right half is sorted
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1, comparisons

bin_index, bin_comparisons = rotated_binary_search(rotated_dataset, target_incident)

# 5. Linear Search Implementation & Comparison Count
def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

lin_index, lin_comparisons = linear_search(rotated_dataset, target_incident)

# Assessment Data Output - Exercise 1
print("Assessment Data:")
print(f"1. Original Dataset: {original_dataset}")
print(f"2. Rotated dataset: {rotated_dataset} (Rotation pivot: {rotation_index})")
print(f"3. Target incident: {target_incident}")
print(f"4. Comparison counts:")
print(f"   - Rotated Binary Search Comparisons: {bin_comparisons} (Found at index {bin_index})")
print(f"   - Linear Search Comparisons: {lin_comparisons} (Found at index {lin_index})")
print(f"5. Time and space complexity:")
print(f"   - Rotated Binary Search -> Time: O(log n), Space: O(1)")
print(f"   - Linear Search -> Time: O(n), Space: O(1)")
print(f"6. Final Output: Target {target_incident} successfully located at index {bin_index} in rotated dataset.\n")
```
